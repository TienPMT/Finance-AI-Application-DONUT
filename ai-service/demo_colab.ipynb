```
import pandas as pd
import numpy as np
from itertools import combinations
from collections import defaultdict
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, mean_absolute_error

print("Đã import thành công các thư viện cần thiết!")
```

## 1. Thuật toán Apriori (Khai thác luật kết hợp)

```
def get_frequent_1_itemsets(transactions, min_sup_count):
    counts = defaultdict(int)
    for basket in transactions:
        for item in basket:
            counts[frozenset([item])] += 1
    return {i: c for i, c in counts.items() if c >= min_sup_count}

# Dữ liệu giả lập demo nhanh (Giỏ hàng)
data = [
    ["Sữa", "Bỉm", "Bia"],
    ["Sữa", "Bỉm", "Bánh mì"],
    ["Sữa", "Bánh mì", "Bơ"],
    ["Bỉm", "Bia"],
    ["Sữa", "Bỉm", "Bia", "Bánh mì"]
]

min_sup_count = 2
f1 = get_frequent_1_itemsets(data, min_sup_count)
print("Tập phổ biến 1-item:", [list(x) for x in f1.keys()])
```

## 2. Hồi quy tuyến tính (Dự báo chi tiêu)

```
# Tính năng: Dự báo số tiền chi tiêu tháng tới dựa trên 3 tháng trước
# X: [Tháng N-3, Tháng N-2, Tháng N-1], Y: Tháng N
X = np.array([[10, 12, 11], [5, 6, 5], [20, 18, 22], [8, 9, 8], [15, 16, 15]])
y = np.array([12, 6, 25, 9, 16])

model_reg = LinearRegression()
model_reg.fit(X, y)
y_pred = model_reg.predict(X)

print("Hệ số (Coefficients):", model_reg.coef_)
print("Dự báo thử:", y_pred)
print("Sai số MAE:", mean_absolute_error(y, y_pred))
```

## 3. Phân lớp khách hàng (Decision Tree)

```
# Features: [Tần suất mua, Tổng loại mặt hàng, Tổng tiền TB/ngày]
# Labels: 1 (VIP - Chi cao), 0 (Thường)
X_cls = np.array([[10, 5, 200], [2, 1, 50], [15, 8, 300], [1, 2, 40], [12, 6, 250]])
y_cls = np.array([1, 0, 1, 0, 1])

clf = DecisionTreeClassifier(max_depth=3)
clf.fit(X_cls, y_cls)

test_cust = np.array([[3, 2, 60], [14, 7, 280]])
print("Dự đoán lớp (0: Thường, 1: VIP):", clf.predict(test_cust))
```
